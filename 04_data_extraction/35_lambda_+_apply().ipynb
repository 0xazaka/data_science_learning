```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('players_20.csv')

# set index
df.set_index('short_name', inplace=True)

# select necessary columns
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club', 'overall']]

# show dataframe
df.head()
```

### 1. Lambda function

its useful when we want to create a temporary function. i.e, we use the function only once

```python
# basic function (that sums two numbers)

def sum_values(a, b):
    x = a + b
    return x

sum_values(2,3)
```

```python
# lambda function (one liner)
sum_values_lambda = lambda a,b:a+b

sum_values_lambda(2,3)
```

### 2. Apply + Lambda Function

```python
# use the lambda function to convert 'height_cm' series to meters
df['height_cm'].apply(lambda x: x/100)

# alternative
df['height_cm']/100
```

```python
# use the lambda function to convert 'long name' series to upper case
df['long_name'].apply(lambda x: x.upper())

# alternative with str attribute
df['long_name'].str.upper()
```

```python
df['dob'] = df['dob'].astype('datetime64[ns]')

# use the lambda function to get the year or 'dob' series
df['dob'].apply(lambda x:x.year)

# alternative with dt attribute
df['dob'].dt.year
```

```python
# apply lambda function to the dataframe in order to calculate the BMI

df.apply(lambda x:x['weight_kg'] / ((x['height_cm']/100)**2), axis=1)
```
