```python
import pandas as pd
```

```python
# read csv file
df = pd.read_csv('players_20.csv')

# set index
df.set_index('short_name', inplace=True)

# select necessary columns
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club', 'overall']]

# show dataframe
df.head()
```

### 1. Drop Row

```python
# drop one row using:

# 1. axis parameter
df.drop('L. Messi', axis=0)

# OR

# 2. index parameter
df.drop(index=['L. Messi'])
```

```python
# drop two or more rows and update data (inplace=True)

df.drop(['L. Messi', 'Neymar Jr'], inplace=True)
df.head()
```

### 1. Drop Columns

```python
# drop one column using:
# 1. axis parameter
df.drop('age', axis=1)

# OR

# 2. columns parameter
df.drop(columns=['age'])
```

### 2. Drop columns by position/index

```python
# drop one column
# axis parameter

df.drop('long_name', axis=1)

# or 
# column parameter
df.drop(columns=['long_name'])
```

```python
# drop columns by position/index (drop last column)


df.drop(df.columns[[-2]], axis=1)
```

```python
# drop two or more columns and update data (inplace=True)

df.drop(['long_name', 'dob'], axis=1, inplace=True)
df
```
