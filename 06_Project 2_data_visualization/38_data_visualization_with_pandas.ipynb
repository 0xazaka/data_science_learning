```python
import pandas as pd
```

```python
# read csv file
df_population_raw = pd.read_csv('population_total.csv')
df_population_raw.head()
```

### 1. Making a pivot table
.pivot(): Returns reshaped DataFrame organized by given index/column values ('Pivot without aggregation')

```python
# dropping null values

df_population_raw.dropna(inplace=True)
```

```python
# making a pivot table

df_pivot = df_population_raw.pivot(index = 'year', columns = 'country', values = 'population')
df_pivot
```

```python
# select some countries 

df_pivot = df_pivot[['Nigeria', 'Japan', 'New Zealand', 'United States', 'Iran', 'Colombia', 'Spain']]
```

```python
# showing pivot table
df_pivot
```

### 2. Lineplot

```python
df_pivot.plot(kind='line',
              xlabel='Year',
              ylabel='Population',
              title='Population from 1955 to 2020',
              figsize=(8,4))
```

### 3. Barplot

```python
# select a year and show the dataframe

df_pivot_2020 = df_pivot.loc[[2020]]
df_pivot_2020
```

```python
# transpose df from rows to columns or vice versa
df_pivot_2020 = df_pivot_2020.transpose()
df_pivot_2020
```

```python
# make barplot
df_pivot_2020.plot(kind='bar', color='green', xlabel='Year', ylabel='Population', title='Population 2020')
```

### 3.2 Barplot grouped by 'n' variables

```python
# select a few years

df_pivot_sample = df_pivot.loc[[1980, 1990, 2000, 2010, 2020]]
df_pivot_sample
```

```python
df_pivot_sample.plot(kind='bar', xlabel='Year', ylabel='Population')
```

### 4. Pie chart

```python
# change column name
df_pivot_2020.rename(columns={2020:'2020'}, inplace=True)
df_pivot_2020
```

```python
# making the plot
df_pivot_2020.plot(kind='pie', y='2020', title='Population in 2020(%)')
```
