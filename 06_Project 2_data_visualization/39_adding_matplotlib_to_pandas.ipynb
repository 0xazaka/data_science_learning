```python
import pandas as pd
import matplotlib.pyplot as plt

```

```python
# read csv file
df_population_raw = pd.read_csv('population_total.csv')
df_population_raw.head()
```

### 1. Making pivot table

```python
# dropping null values

df_population_raw.dropna(inplace=True)
```

```python
# making a pivot table

df_pivot = df_population_raw.pivot(index = 'year', columns = 'country', values = 'population')
df_pivot
```

```python
# select some countries and show pivot table

df_pivot = df_pivot[['Nigeria', 'Japan', 'New Zealand', 'United States', 'Iran', 'Colombia', 'Spain']]
df_pivot
```

### 2. Lineplot

```python
df_pivot.plot(kind='line',
              xlabel='Year',
              ylabel='Population',
              title='Population from 1955 to 2020',
              figsize=(8,4))


# saving the plot
plt.savefig('my_test.png')

```

```python
# exporting to excel

df_pivot.to_excel('pivot_table.xlsx')
```
