```python
import pandas as pd
```

```python
df_movies = pd.read_csv("IMDb movies.csv", low_memory=False)
df_ratings = pd.read_csv("IMDb ratings.csv")
```

```python
# select columns

df_movies = df_movies[['imdb_title_id', 'title', 'year', 'genre', 'country']]
df_ratings = df_ratings[['imdb_title_id', 'total_votes', 'mean_vote']]
```

### 1. merge()

```python
df1 = pd.DataFrame({
    'id': ['A','B','C','D'],
    'age': [30, 23, 25, 22]
})
df2 = pd.DataFrame({
    'id': ['C','D','E','F'],
    'job': ['Doctor', 'Statistician', 'Accountant', 'Developer']
})
```

### 1.1 Inner Join

```python
df_inner_join = df1.merge(df2, on='id', how='inner')
df_inner_join
```

```python
#shape
print(df1.shape)
print(df2.shape)
print(df_inner_join.shape)
```

### 1.2 Outer join (Full join)

```python
df_outer_join = df1.merge(df2, on='id', how='outer')
df_outer_join
```

```python
# shape

print(df1.shape)
print(df2.shape)
print(df_inner_join.shape)
print(df_outer_join.shape)
```

### 1.3 Exclusive Outer join (Exclusive full join)

```python
df1.merge(df2, on='id', how='outer', indicator=True)
```

```python
# query

df_exclusive_outer_join = df1.merge(df2, on='id', how='outer', indicator=True).query("_merge=='left_only' or _merge=='right_only'")
df_exclusive_outer_join
```

```python
# shape
print(df_exclusive_outer_join.shape)
```
