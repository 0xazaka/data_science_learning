### 1 concat()
#### 1.1 Concatenate vertically

```python
df1 = pd.DataFrame({
    'id': ['A','B','C','D'],
    'age': [30, 23, 25, 22]
})
df2 = pd.DataFrame({
    'id': ['E','F','G','H'],
    'age': [40, 21, 19, 24]
})
```

```python
pd.concat([df1, df2], ignore_index=True) 

# axis = 0 by default
# "ignore_index" ignores the index of the next dataframe
```
